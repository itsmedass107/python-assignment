```python
import google.generativeai as genai
from google.colab import userdata

genai.configure(api_key=userdata.get("GEMINI_API_KEYS"))

knowledge_base = """
FitZone Gym Knowledge Base

1. Gym timings are 5:00 AM to 10:00 PM every day.
2. Monthly membership fee is ₹1500.
3. Yearly membership fee is ₹15000.
4. Registration fee is ₹500.
5. Certified personal trainers are available.
6. Beginners are welcome.
7. Free parking is available.
8. Lockers are available for members.
9. Diet plans are provided with personal training.
10. Group classes include Yoga, Zumba and HIIT.
11. Cardio equipment like treadmills and cycles are available.
12. Membership can be frozen for one month.
13. We accept Cash, UPI, Debit Card and Credit Card.
14. Bring gym shoes, towel and water bottle on your first day.
15. Contact Number: 9876543210
"""

# Create Model
model = genai.GenerativeModel(
    model_name="gemini-2.5-flash-lite",

    system_instruction=f"""
You are the FAQ assistant for FitZone Gym.

Use ONLY the information given below.

Knowledge Base:
{knowledge_base}

Rules:
- Answer only questions related to FitZone Gym.
- If the answer is not available in the knowledge base, reply:
"Sorry, I can answer only questions related to FitZone Gym."
- Keep answers short and simple.
"""
)
chat = model.start_chat(history=[])

print("====================================")
print("     FITZONE GYM FAQ CHATBOT")
print("====================================")
print("Type 'quit' or 'exit' to stop.\n")

while True:

    question = input("You: ")

    if question.lower() in ["quit", "exit"]:
        print("\nBot: Thank you for visiting FitZone Gym!")
        break

    response = chat.send_message(
        question,
        generation_config={
            "temperature": 0.2
        }
    )

    print("\nBot:", response.text)
    print()

    # Save Chat History
    with open("gym_chat_history.txt", "a") as file:
        file.write("User: " + question + "\n")
        file.write("Bot: " + response.text + "\n\n")
```

```python
import google.generativeai as genai
from google.colab import userdata

# Configure API
genai.configure(api_key=userdata.get("GEMINI_API_KEYS"))

# Create Model
model = genai.GenerativeModel(
    model_name="gemini-2.5-flash-lite",

    system_instruction="""
You are a professional email writer.

Rules:
- Write a proper subject.
- Write a polite email.
- End with a professional closing.
- Follow the user's requested tone.
"""
)

print("===== EMAIL GENERATOR =====")
print("Type 'quit' to exit.\n")

while True:

    recipient = input("Recipient: ")

    if recipient.lower() == "quit":
        print("Goodbye!")
        break

    purpose = input("Purpose: ")
    tone = input("Tone (Formal/Casual/Professional): ")

    prompt = f"""
Recipient: {recipient}

Purpose: {purpose}

Tone: {tone}
"""

    response = model.generate_content(
        prompt,
        generation_config={"temperature":0.3}
    )

    print("\n===== GENERATED EMAIL =====\n")
    print(response.text)

    with open("email_output.txt","w") as file:
        file.write(response.text)

    print("\nEmail saved as email_output.txt\n")
```
